```python
import numpy as np
import matplotlib.pyplot as plt
import linecache

Msolar = 1.989000e+33
G = 6.67408e-08
AU = 1.495979e+13
kB = 1.38064852e-16
mH = 1.67353e-24
mu = 2.34
yr = 3.155760e+07
```

```python
filename = "test.txt"
species = linecache.getline(filename, 1)
species = species.split()[1:]
data = np.loadtxt(filename, float, skiprows=1).T
t = data[0]
abunds = data[1:]
# abunds.astype(np.float)
del data
print(species)
t = t /yr
```

```python
plt.figure(figsize=(12, 7))
for i in range(len(species)):
    # if not (("+" in species[i]) or ("-" in species[i])): # or (species[i][0] == "G"):
    if (species[i] == "C"):
        plt.plot(t, abunds[i], label=species[i], linewidth=4, color="black")
    elif (species[i] == "CO"):
        plt.plot(t, abunds[i], label=species[i], linewidth=4)
    elif (species[i] == "H2"):
        print(abunds[i][-1])
        plt.plot(t, abunds[i], label=species[i])
    else:
        plt.plot(t, abunds[i], label=species[i])
        pass
plt.xscale("log")
plt.yscale("log")
# plt.xlim(1.0)
plt.ylim(1.0e-12, 1.0e0)
# plt.legend(fontsize=15, loc='upper left', bbox_to_anchor=(1, 1))
plt.tick_params(labelsize=20)
# plt.xticks([10.0**i for i in [-6,  -4, -2, 0, 2, 4, 6]])
plt.xlabel("year", fontsize=20)
plt.ylabel("abundance", fontsize=20)
plt.legend(fontsize=15, loc='upper center', bbox_to_anchor=(.5, -.15), ncol=5)
plt.show()

linecache.clearcache()
```
